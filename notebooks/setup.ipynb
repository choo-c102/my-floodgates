# Set Up for Cloud Integration

**Before we start:**
This notebook requires the working IDE to have 'AWS Toolkit for VS Code' extension installed - you can find this from the marketplace. Sign-in to AWS using personal credentials.

To remotely connect (SSH) VS Code to AWS resources, you need an SSH client. Firstly, make sure the 'Remote- SSH' and 'Remote - SSH:Editing Configuration' extensions are installed. Secondly, ensure that an SSH client is installed on your local device. Most modern OS should already have an SSH client pre-installed, but it doesn't hurt to check!

When you're creating your AWS resources (S3 buckets, EC3 instances, etc),  ensure the following:
- the instances have passed their status check
- save the .pem keys somewhere safe
- allow inbound SSH traffic from your IP address under your security group
